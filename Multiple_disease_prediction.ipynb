```python
# import library

import pandas as pd
```

```python
# read datset

df = pd.read_csv('https://github.com/YBIFoundation/Dataset/raw/main/MultipleDiseasePrediction.csv')
```

```python
# define x and y

x = df.drop(['prognosis'], axis=1)

y = df['prognosis']
```

```python
# train test split

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

```

```python
# model selection

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
log_reg = LogisticRegression()
knn = KNeighborsClassifier()
dt = DecisionTreeClassifier()
rf = RandomForestClassifier()
```

```python
# train model

log = log_reg.fit(x_train, y_train)
knn = knn.fit(x_train, y_train)
dt = dt.fit(x_train, y_train)
rf = rf.fit(x_train, y_train)
```

```python
# prediction

pred_log = log.predict(x_test)
pred_knn = knn.predict(x_test)
pred_dt = dt.predict(x_test)
pred_rf = rf.predict(x_test)
```

```python
# evaluation

from sklearn.metrics import  classification_report

cr1 = classification_report(y_test, pred_log)
cr2 = classification_report(y_test, pred_knn)
cr3 = classification_report(y_test, pred_dt)
cr4 = classification_report(y_test, pred_rf)

print("Logistic Regression\n")
print(cr1)
print("______________________***********_________________________\n")


print("KNN\n")
print(cr2)
print("______________________***********_________________________\n")


print("Decision Tree\n")
print(cr3)
print("______________________***********_________________________\n")


print("Random Forest\n")
print(cr4)
print("______________________***********_________________________\n")

```
